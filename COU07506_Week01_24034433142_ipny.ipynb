```python
# Standard imports
import numpy as np
import matplotlib.pyplot as plt

# Title
# Elementary Sequences

n = np.arange(-5, 11)
delta = (n == 0).astype(float)   # unit impulse
step = (n >= 0).astype(float)     # unit step
ramp = n * step                   # unit ramp
expo = 0.8**n * step              # real exponential

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))
for ax, x, title in zip(axes, [delta, step, ramp, expo],
                        [r"$\delta(n)$", "u(n)", "r(n)", r"$(0.8)^n u(n)$"]):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title); ax.set_xlabel("n")
plt.tight_layout(); plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import librosa
import os

fname = "my_voice.wav (4).mp4"   # change to your file name
if os.path.exists(fname):
    # librosa handles MP4, M4A, WAV, and MP3 formats natively
    x, fs = librosa.load(fname, sr=None)
    # Normalize the audio signal to [-1, 1]
    x = x / np.max(np.abs(x))
else:  # fallback test signal
    fs = 16000
    t = np.arange(0, 2, 1/fs)
    x = 0.5*np.sin(2*np.pi*440*t) * np.exp(-t)

t = np.arange(len(x)) / fs
print(f"fs = {fs} Hz, duration = {len(x)/fs:.2f} s, samples = {len(x)}")
plt.figure(figsize=(10, 4))
plt.plot(t, x)
plt.xlabel("time (s)")
plt.ylabel("amplitude")
plt.title("My recording")
plt.grid(True)
plt.show()
```
